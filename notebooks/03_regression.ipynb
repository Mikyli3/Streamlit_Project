```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
```

```python
base_features = final_features.copy()

print(base_features)
```

```python
X_train = df_train[base_features]
X_val = df_val[base_features]

y_train = np.log1p(df_train["positive"])
y_val = np.log1p(df_val["positive"])
```

```python
dummy = DummyRegressor(strategy="mean")

dummy.fit(X_train, y_train)

dummy_pred = dummy.predict(X_val)

dummy_mae = mean_absolute_error(y_val, dummy_pred)
dummy_r2 = r2_score(y_val, dummy_pred)

print("MAE:", dummy_mae)
print("R²:", dummy_r2)
```

```python
model = LinearRegression()

model.fit(X_train, y_train)

model_pred = model.predict(X_val)

model_mae = mean_absolute_error(y_val, model_pred)
model_r2 = r2_score(y_val, model_pred)

print("MAE:", model_mae)
print("R²:", model_r2)
```

```python
baseline_results = pd.DataFrame({
    "Model": [
        "DummyRegressor",
        "LinearRegression"
    ],
    "MAE": [
        dummy_mae,
        model_mae
    ],
    "R²": [
        dummy_r2,
        model_r2
    ]
})

baseline_results.round(4)
```

## 2. Kandidater

```python
genre_counts = (
    df_train["genres"]
    .explode()
    .value_counts()
)

category_counts = (
    df_train["categories"]
    .explode()
    .value_counts()
)

display(genre_counts)
display(category_counts)
```

```python
min_occurrences = 1000

common_genres = genre_counts[
    genre_counts >= min_occurrences
].index.tolist()

common_categories = category_counts[
    category_counts >= min_occurrences
].index.tolist()

print("Genres:", common_genres)
print("Categories:", common_categories)
```

```python
pd.DataFrame({
    "Category": common_categories
})
```

```python
excluded_categories = [
    "Steam Trading Cards"
]

common_categories = [
    c for c in common_categories
    if c not in excluded_categories
]

candidates = []

for genre in common_genres:

    column = f"genre_{genre}"

    for d in [df_train, df_val, df_test]:
        d[column] = d["genres"].apply(
            lambda g: int(genre in g)
        )

    candidates.append(column)


for category in common_categories:

    column = f"category_{category}"

    for d in [df_train, df_val, df_test]:
        d[column] = d["categories"].apply(
            lambda c: int(category in c)
        )

    candidates.append(column)

print("Number of candidates:", len(candidates))
```

```python
for d in [df_train, df_val, df_test]:

    d["is_free"] = (d["price"] == 0).astype(int)

    d["required_age"] = d["required_age"].fillna(0)

    d["n_categories"] = d["categories"].apply(len)

    d["n_genres"] = d["genres"].apply(len)

    d["n_audio_languages"] = (
        d["full_audio_languages"].apply(len)
    )

    d["log_price"] = np.log1p(d["price"])

candidates.extend([
    "is_free",
    "required_age",
    "n_categories",
    "n_genres",
    "n_audio_languages",
    "log_price"
])

candidates = list(dict.fromkeys(candidates))

candidates = [
    c for c in candidates
    if c not in base_features
]
```

```python
correlations = (
    df_train[candidates]
    .corrwith(y_train)
    .dropna()
)

correlation_table = pd.DataFrame({
    "Correlation": correlations,
    "Absolute correlation": correlations.abs()
})

correlation_table = correlation_table.sort_values(
    "Absolute correlation",
    ascending=False
)

correlation_table.round(4)
```

## 3. En feature i taget

```python
selected_features = base_features.copy()

best_r2 = model_r2
best_mae = model_mae

min_improvement = 0.002

selection_results = []

for candidate in correlation_table.index:

    trial_features = selected_features + [candidate]

    trial_model = LinearRegression()

    trial_model.fit(
        df_train[trial_features],
        y_train
    )

    predictions = trial_model.predict(
        df_val[trial_features]
    )

    mae = mean_absolute_error(y_val, predictions)
    r2 = r2_score(y_val, predictions)

    improvement = r2 - best_r2

    keep = improvement >= min_improvement

    selection_results.append({
        "Feature": candidate,
        "MAE": mae,
        "R²": r2,
        "R² improvement": improvement,
        "Retained": keep
    })

    if keep:
        selected_features.append(candidate)
        best_r2 = r2
        best_mae = mae

selection_table = pd.DataFrame(selection_results)

selection_table.round(4)
```

```python
selection_table[
    selection_table["Retained"]
].round(4)
```

```python
final_model = LinearRegression()

final_model.fit(
    df_train[selected_features],
    y_train
)

final_pred = final_model.predict(
    df_val[selected_features]
)

final_mae = mean_absolute_error(y_val, final_pred)
final_r2 = r2_score(y_val, final_pred)

print("Selected features:", selected_features)
print("MAE:", final_mae)
print("R²:", final_r2)
```

```python
regression_results = pd.DataFrame({
    "Model": [
        "DummyRegressor",
        "Starting model",
        "Improved model"
    ],
    "MAE": [
        dummy_mae,
        model_mae,
        final_mae
    ],
    "R²": [
        dummy_r2,
        model_r2,
        final_r2
    ]
})

regression_results.round(4)
```

```python
regression_results.set_index("Model")["R²"].plot(
    kind="bar",
    figsize=(10, 5),
    rot=0
)

plt.title("Model comparison – R²")
plt.ylabel("R²")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## Slutsats och motivering av den slutliga modellen

Syftet med milstolpe 3 var att förutspå antalet positiva recensioner med hjälp av linjär regression. Vi använde det rensade datat från milstolpe 2 och logaritmerade målvariabeln eftersom antalet positiva recensioner har en mycket sned fördelning.

### Resultat

| Modell | MAE | R² |
|---|---:|---:|
| DummyRegressor | 1,8934 | −0,0001 |
| Startmodell | 1,6421 | 0,2272 |
| Modell före korrigering | 1,5072 | 0,3527 |
| Korrigerad modell | 1,5735 | 0,3024 |

Vi testade nya features systematiskt genom att lägga till en feature i taget. En feature behölls endast om R² ökade med minst 0,002 på valideringsdatat.

Under arbetet identifierade vi Steam Trading Cards som en potentiell källa till dataläckage, eftersom tillgången till samlarkort kan påverkas av spelets framgång efter lanseringen. Vi valde därför att exkludera denna feature och genomföra feature-selektionen på nytt.

Trots att R² sjönk efter korrigeringen förbättrades resultatet jämfört med startmodellen. R² ökade från 0,2272 till 0,3024, medan MAE minskade med cirka 4,2 %.

### De största felen

Modellen har begränsad information om faktorer som marknadsföring, spelkvalitet, popularitet på sociala medier och spelarengagemang. Dessa faktorer kan ha stor betydelse för antalet positiva recensioner men är svåra att förutspå innan ett spel har lanserats.

### Slutlig bedömning

Den korrigerade modellen förklarar cirka 30,2 % av variationen i antalet positiva recensioner på log-skalan. Resultatet ligger nära det förväntade värdet i uppgiften.

Vi föredrar den korrigerade modellen framför modellen med högre R² eftersom den minskar risken för dataläckage. En tillförlitlig modell är viktigare än ett högre R² som bygger på information som kanske inte är tillgänglig före lanseringen.

Den slutliga utvärderingen på testdatat genomförs efter att gruppen har granskat och godkänt modellens features.
