# Milstolpe 1: Första modellen

**Mina features: hur påkostat och genomarbetat spelet ser ut på butikssidan**

Robin valde pris, plattformar och antal språk. Jag väljer i stället features som säger något om **hur stort och påkostat projektet är**. Ett stort spel har ofta en lång beskrivning, många skärmbilder, en egen webbplats, achievements och planerat DLC. Ett litet hobbyspel har oftast inget av det.

| Feature | Hur den räknas | Varför |
|---|---|---|
| `dlc_count` | antal DLC | Planerat DLC betyder ofta en större studio och ett spel som väntas sälja |
| `achievements` | antal achievements | Visar hur mycket arbete som lagts på spelet |
| `n_screenshots` | antal skärmbilder på butikssidan | En genomarbetad butikssida lockar fler köpare |
| `n_movies` | antal trailers | Trailer = marknadsföringsbudget |
| `desc_length` | antal tecken i `detailed_description` | En lång beskrivning tyder på ett större och mer genomarbetat spel |
| `has_website` | 1 om spelet har en webbplats, annars 0 | En egen webbplats = en mer seriös utgivare |

Alla dessa finns på butikssidan **före** lanseringen. `dlc_count` och `achievements` är en gråzon: de kan också läggas till efter lanseringen, när spelet redan har sålt bra. Det kontrollerar vi närmare i milstolpe 2.

```python
# Gör om kolumnerna till tal
for d in [df_train, df_val, df_test]:
    d["n_screenshots"] = d["screenshots"].apply(len)
    d["n_movies"] = d["movies"].apply(len)
    d["desc_length"] = d["detailed_description"].str.len()
    d["has_website"] = (d["website"].str.strip() != "").astype(int)

features = ["dlc_count", "achievements", "n_screenshots", "n_movies", "desc_length", "has_website"]

df_train[features].describe().T
```

`n_movies` är **0 för alla spel**: datasetet innehåller inga trailers. Featuren kan alltså inte hjälpa modellen. Jag behåller den ändå i testerna nedan för att visa det.

```python
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, recall_score

X_train = df_train[features]
y_train = df_train["hit"]
X_val = df_val[features]
y_val = df_val["hit"]

# Lat jämförelsemodell: gissar alltid "inte hit"
dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train, y_train)
gissningar = dummy.predict(X_val)

print("Dummy")
print("Accuracy:", round(accuracy_score(y_val, gissningar), 4))
print("Recall:", round(recall_score(y_val, gissningar), 4))
```

```python
# Logistisk regression med alla mina features
modell = LogisticRegression(max_iter=3000)
modell.fit(X_train, y_train)
gissningar = modell.predict(X_val)

print("Logistisk regression")
print("Accuracy:", round(accuracy_score(y_val, gissningar), 4))
print("Recall:", round(recall_score(y_val, gissningar), 4))
```

Modellen slår baseline: accuracy är lite högre och den hittar ungefär 11 % av alla hits (baseline hittar 0 %).

## En feature i taget
Först tar jag bort en feature i taget. Om recall inte sjunker när en feature tas bort, hjälper den inte. Sedan tränar jag modellen med bara en feature åt gången.

```python
import pandas as pd

def testa(kolumner, class_weight=None):
    m = LogisticRegression(max_iter=3000, class_weight=class_weight)
    m.fit(df_train[kolumner], y_train)
    p = m.predict(df_val[kolumner])
    return round(accuracy_score(y_val, p), 4), round(recall_score(y_val, p), 4)

rader = []
rader.append(["alla features", *testa(features)])
for f in features:
    utan = [x for x in features if x != f]
    rader.append(["utan " + f, *testa(utan)])
for f in features:
    rader.append(["bara " + f, *testa([f])])

pd.DataFrame(rader, columns=["modell", "accuracy", "recall"])
```

**Vad hjälper inte?**
- `n_movies` hjälper inte alls. Resultatet blir exakt detsamma utan den, eftersom den alltid är 0.
- `achievements` hjälper bara lite: recall sjunker från 0,111 till 0,108 utan den.
- `dlc_count` och `desc_length` hjälper mest: utan dem sjunker recall till ungefär 0,08.
- `has_website` och `n_screenshots` hjälper lite.

## Extra: ge hits mer vikt
Hits är bara 25 % av spelen, så modellen gissar oftast "inte hit". Med `class_weight="balanced"` räknas ett missat hit som ett större fel. Robin provade samma sak, så vi kan jämföra resultaten rättvist.

```python
modell_balans = LogisticRegression(max_iter=3000, class_weight="balanced")
modell_balans.fit(X_train, y_train)
gissningar = modell_balans.predict(X_val)

print("Logistisk regression, balanced")
print("Accuracy:", round(accuracy_score(y_val, gissningar), 4))
print("Recall:", round(recall_score(y_val, gissningar), 4))

rader = [["alla features", *testa(features, "balanced")]]
for f in features:
    utan = [x for x in features if x != f]
    rader.append(["utan " + f, *testa(utan, "balanced")])
pd.DataFrame(rader, columns=["modell", "accuracy", "recall"])
```

- Features som visar hur påkostat ett spel är hittar fler hits än baseline.
- `n_movies` saknar information i datasetet och kan tas bort. `achievements` hjälper bara lite.
- Med `class_weight="balanced"` hittar modellen ungefär 59 % av alla hits, men gör fler falska alarm, så accuracy sjunker.
- `dlc_count` och `achievements` kan ha ändrats efter lanseringen (gråzon). Det undersöker vi i milstolpe 2.
