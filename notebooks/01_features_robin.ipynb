# Milstolpe 1 – Robin

Jag valde saker som påverkar hur många som kan köpa spelet. Ett billigt spel som funkar på många datorer och språk borde nå fler spelare och få fler bra recensioner.

| Feature | Varför |
|---|---|
| `price` | Billigare spel är lättare att köpa |
| `windows`, `mac`, `linux` | Fler plattformar = fler som kan spela |
| `antal_sprak` | Folk köper hellre spel på sitt eget språk |

Allt det här syns på Steam-sidan innan spelet släpps. Men priset kan ändras med rea, och språk kan läggas till senare. Det kollar vi i milstolpe 2.

```python
# Egenskaper som modellen ska använda
features = ["price", "windows", "mac", "linux"]

# Visar några exempel från träningsdatan
df_train[features].head()
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, recall_score

# Väljer information och facit
X_train = df_train[features]
y_train = df_train["hit"]

X_val = df_val[features]
y_val = df_val["hit"]

# Tränar och testar jämförelsemodellen
dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train, y_train)
gissningar = dummy.predict(X_val)

print("Accuracy:", accuracy_score(y_val, gissningar))
print("Recall:", recall_score(y_val, gissningar))
```

```python
from sklearn.linear_model import LogisticRegression

# Tränar modellen
modell = LogisticRegression(max_iter=1000)
modell.fit(X_train, y_train)

# Testar på valideringsdatan
gissningar = modell.predict(X_val)

print("Accuracy:", accuracy_score(y_val, gissningar))
print("Recall:", recall_score(y_val, gissningar))
```

```python
# Testar att ta bort en egenskap i taget
for egenskap in features:
    utan = [f for f in features if f != egenskap]

    modell_test = LogisticRegression(max_iter=1000)
    modell_test.fit(df_train[utan], y_train)

    gissningar = modell_test.predict(df_val[utan])

    print("Utan", egenskap)
    print("Accuracy:", round(accuracy_score(y_val, gissningar), 4))
    print("Recall:", round(recall_score(y_val, gissningar), 4))
    print()
```

```python
# Kopierar de tidigare egenskaperna
X_train_sprak = X_train.copy()
X_val_sprak = X_val.copy()

# Räknar antal språk för varje spel
X_train_sprak["antal_sprak"] = df_train["supported_languages"].apply(len)
X_val_sprak["antal_sprak"] = df_val["supported_languages"].apply(len)

# Tränar en ny modell
modell_sprak = LogisticRegression(max_iter=1000)
modell_sprak.fit(X_train_sprak, y_train)

gissningar_sprak = modell_sprak.predict(X_val_sprak)

# Skriver ut resultat
print("Accuracy:", accuracy_score(y_val, gissningar_sprak))
print("Recall:", recall_score(y_val, gissningar_sprak))
```

```python
# Ger hits mer vikt
modell_balans = LogisticRegression(max_iter=1000, class_weight="balanced")
modell_balans.fit(X_train_sprak, y_train)

gissningar_balans = modell_balans.predict(X_val_sprak)

print("Accuracy:", accuracy_score(y_val, gissningar_balans))
print("Recall:", recall_score(y_val, gissningar_balans))
```

```python
# Testar att ta bort en egenskap i taget
for egenskap in X_train_sprak.columns:
    utan = X_train_sprak.columns.drop(egenskap)

    testmodell = LogisticRegression(max_iter=1000, class_weight="balanced")
    testmodell.fit(X_train_sprak[utan], y_train)

    gissningar = testmodell.predict(X_val_sprak[utan])

    print("Utan", egenskap)
    print("Accuracy:", round(accuracy_score(y_val, gissningar), 4))
    print("Recall:", round(recall_score(y_val, gissningar), 4))
    print()
```
