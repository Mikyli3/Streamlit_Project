# Milstolpe 4: Lovande eller inte?

Appen ska säga LOVANDE eller REKOMMENDERAS INTE. Vi använder det rensade datat från milstolpe 2 och de bästa featurena från milstolpe 3.

```python
import numpy as np
import pandas as pd

# Features som valdes i milstolpe 3
for d in [df_train, df_val, df_test]:
    d["n_categories"] = d["categories"].apply(len)
    d["is_free"] = (d["price"] == 0).astype(int)
    d["genre_Casual"] = d["genres"].apply(lambda g: int("Casual" in g))
    for kat in ["Steam Cloud", "Remote Play on TV", "Steam Workshop", "Remote Play Together", "Shared/Split Screen"]:
        d["category_" + kat] = d["categories"].apply(lambda c: int(kat in c))

features = [
    "log_dlc", "log_achievements", "n_screenshots", "log_desc_length", "has_website",
    "n_categories", "category_Steam Cloud", "category_Remote Play on TV", "required_age",
    "category_Steam Workshop", "genre_Casual", "is_free",
    "category_Remote Play Together", "category_Shared/Split Screen",
]

df_train[features].head()
```

## 1. Bättre mått

Accuracy räcker inte, eftersom bara 25 % av spelen är hits. Därför tittar vi på:
- **Precision:** när modellen säger "hit", hur ofta stämmer det?
- **Recall:** av alla riktiga hits, hur många hittar modellen?
- **F1:** ett tal som väger ihop precision och recall.

```python
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score

X_train = df_train[features]
y_train = df_train["hit"]
X_val = df_val[features]
y_val = df_val["hit"]

dummy = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
modell = LogisticRegression(max_iter=3000).fit(X_train, y_train)
modell_balans = LogisticRegression(max_iter=3000, class_weight="balanced").fit(X_train, y_train)

rader = []
for namn, m in [("Dummy", dummy), ("Logistisk regression", modell), ("Logistisk regression, balanced", modell_balans)]:
    p = m.predict(X_val)
    rader.append([
        namn,
        precision_score(y_val, p, zero_division=0),
        recall_score(y_val, p),
        f1_score(y_val, p),
    ])

pd.DataFrame(rader, columns=["modell", "precision", "recall", "F1"]).round(3)
```

**Förväxlingsmatris**

Raderna är sanningen och kolumnerna är vad modellen sa. Uppe till höger är falska alarm, nere till vänster är missade hits.

```python
for namn, m in [("Logistisk regression", modell), ("Logistisk regression, balanced", modell_balans)]:
    cm = confusion_matrix(y_val, m.predict(X_val))
    print(namn)
    display(pd.DataFrame(
        cm,
        index=["Verkligen flopp", "Verkligen hit"],
        columns=["Modellen sa flopp", "Modellen sa hit"],
    ))
```

**Vad betyder det för Pixel Harbor?**

- **Dummy** säger alltid "flopp" och hittar inga hits alls.
- **Vanlig modell:** säger "hit" om ca 2 500 spel. 1 613 av dem är riktiga hits (precision 66 %), men den missar 3 261 hits (recall 33 %). Den är försiktig: få falska alarm, men många missade hits.
- **Balanced:** säger "hit" om ca 6 600 spel. Den hittar 3 138 hits (recall 64 %), men 3 463 av dess "hit" är falska alarm (precision 48 %).
- För Pixel Harbor betyder ett falskt alarm att de finansierar ett spel som floppar. En missad hit betyder att de säger nej till ett spel som hade blivit bra.
- Vilken modell som är bäst beror på hur många spel Pixel Harbor kan finansiera. Det bestämmer vi i del 2 och 3 med ett tröskelvärde.

## 2. Pixel Harbors situation

Våra gissningar:

- Pixel Harbor har bara råd att satsa på ungefär **10 %** av spelidéerna de får.
- **Falskt alarm:** de satsar på ett spel som floppar. Då förlorar de ungefär **200 000 €**.
- **Missad hit:** de säger nej till ett spel som hade blivit en hit. Då missar de ungefär **400 000 €** i vinst. Men pengarna går ändå till ett annat spel.

**Vad betyder det?**
- De kan bara välja några få spel, så de spel modellen säger "hit" om måste verkligen vara bra. Därför är **precision** viktigast.
- Att missa en hit är synd, men de har ändå inte råd med alla spel.
- I del 3 väljer vi ett tröskelvärde där modellen säger "hit" om ungefär 10 % av spelen.

## 3. Välj tröskelvärde

Modellen ger en sannolikhet för hit. Vi provar olika gränser för när den ska säga "hit".

```python
# Sannolikheten för hit för varje spel
proba = modell.predict_proba(X_val)[:, 1]

rader = []
for t in [0.1, 0.15, 0.2, 0.25, 0.3, 0.35, 0.4, 0.45, 0.5, 0.6, 0.7]:
    p = (proba >= t).astype(int)
    rader.append([
        t,
        precision_score(y_val, p, zero_division=0),
        recall_score(y_val, p),
        f1_score(y_val, p),
        p.sum(),
        p.mean(),
    ])

troskel_tabell = pd.DataFrame(rader, columns=["tröskel", "precision", "recall", "F1", "antal sagt hit", "andel sagt hit"])
troskel_tabell.round(3)
```

```python
# Tröskel där modellen säger "hit" om exakt 10 % av spelen
antal = int(len(proba) * 0.10)
troskel_10 = pd.Series(proba).nlargest(antal).min()
p = (proba >= troskel_10).astype(int)

print("Tröskel för 10 %:", round(troskel_10, 3))
print("Precision:", round(precision_score(y_val, p), 3))
print("Recall:", round(recall_score(y_val, p), 3))
print("F1:", round(f1_score(y_val, p), 3))
print("Antal sagt hit:", p.sum())

# Jämförelse: tröskeln med högst F1 i tabellen
basta_f1 = troskel_tabell.loc[troskel_tabell["F1"].idxmax()]
print("\nHögst F1 i tabellen: tröskel", basta_f1["tröskel"], "med F1", round(basta_f1["F1"], 3))
```

**Vilket tröskelvärde väljer vi?**

| Tröskel | Precision | Recall | F1 | Sagt hit |
|---|---|---|---|---|
| 0,25 (högst F1) | 0,46 | 0,67 | 0,55 | 38 % av spelen |
| **0,56 (vårt val)** | **0,70** | **0,27** | **0,39** | **10 % av spelen** |

- Med högst F1 (0,25) säger modellen "hit" om 38 % av spelen. Pixel Harbor har bara råd med 10 %, och mer än hälften av dem skulle floppa.
- Vi väljer **0,56**. Då säger modellen "hit" om 10 % av spelen, och 7 av 10 av dem blir hits. Utan modell blir bara 1 av 4 en hit.
- F1 blir lägre, eftersom vi missar många hits. Men Pixel Harbor har ändå inte råd med fler spel, så det är viktigare att de spel de väljer är bra.

## 4. Testet

Vi testar modellen och tröskeln 0,56 en gång på testdatat.

```python
# Tröskeln som appen ska använda
TROSKEL = 0.56

X_test = df_test[features]
y_test = df_test["hit"]

proba_test = modell.predict_proba(X_test)[:, 1]
p_test = (proba_test >= TROSKEL).astype(int)

print("Precision:", round(precision_score(y_test, p_test), 3))
print("Recall:", round(recall_score(y_test, p_test), 3))
print("Andel sagt hit:", round(p_test.mean(), 3))

pd.DataFrame(
    confusion_matrix(y_test, p_test),
    index=["Verkligen flopp", "Verkligen hit"],
    columns=["Modellen sa flopp", "Modellen sa hit"],
)
```

**Resultat på testdatat**

- Precision 0,71 och recall 0,27. Nästan samma som på valideringsdatat, så modellen fungerar även på nya spel.
- Modellen sa "hit" om 1 824 spel (ca 10 %). 1 297 av dem blev riktiga hits, alltså ungefär 7 av 10.
- 527 var falska alarm, och 3 578 hits missades.

**Appen använder tröskeln 0,56:** sannolikhet 0,56 eller mer ger LOVANDE, annars REKOMMENDERAS INTE.
