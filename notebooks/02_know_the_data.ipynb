## Sifferkolumner

- Många kolumner är nästan alltid 0: `metacritic_score`, `user_score`, `recommendations`, `peak_ccu` och alla playtime-kolumner. Här betyder 0 oftast "saknas". De blir dessutom kända först efter släppet, så de får inte användas.
- `price`, `dlc_count`, `achievements` och `positive` är sneda: de flesta spel har låga värden, men några har jättehöga. Till exempel har `dlc_count` median 0 men max 3703.
- Några värden ser konstiga ut, som ett pris på 999,98 dollar. Det kollar vi nedan.
- `hit` är 0,25 i snitt, alltså är 25 % av spelen hits, som det ska vara.

## Är typen rätt?

- `release_date` var text ("Aug 1, 2023") men görs om till datum i den gemensamma notebooken. Den används ändå inte i modellen.
- `estimated_owners` är text med ett intervall ("0 - 20000"), inte ett tal. Den läcker ändå, så vi rättar den inte.
- `appID` är text, vilket är rätt, eftersom det är ett ID och inget tal att räkna med.
- `windows`, `mac` och `linux` är True/False. Det fungerar, men blir 1/0 när vi gör `.astype(int)`.
- `genres`, `categories`, `supported_languages`, `screenshots` m.fl. är listor. Modellen förstår inte listor, så vi räknar antalet med `.apply(len)`.
- `packages` är text som egentligen är JSON (en lista med priser). Vi använder den inte.
- Sifferkolumnerna (`price`, `dlc_count` …) har rätt typ.

```python
# Vilka spel har konstigt högt pris eller jättemånga DLC?
print("Dyraste spelen:")
display(df_train.nlargest(5, "price")[["name", "price", "positive"]])

print("Flest DLC:")
display(df_train.nlargest(5, "dlc_count")[["name", "dlc_count", "positive"]])
```

**Konstiga värden**
- De dyraste spelen (upp till 999 dollar) är riktiga men udda, och har nästan inga recensioner. Värdena är inte fel, men priset är väldigt snett. Därför behöver det log-skala.
- Spelen med flest DLC (Fantasy Grounds, RPG Maker, Rocksmith, Train Simulator) har fått sina DLC under många år efter släppet, för att de sålde bra. `dlc_count` kan alltså läcka: hits får fler DLC, inte tvärtom.

```python
# Hur många spel har tomma listor eller tom text? (räknas inte som "saknas" av isnull)
for kolumn in ["genres", "categories", "tags", "supported_languages", "developers", "publishers"]:
    tomma = (df_train[kolumn].apply(len) == 0).mean()
    print(f"{kolumn:<22} {tomma:.1%} tomma")

for kolumn in ["short_description", "detailed_description", "website"]:
    tomma = (df_train[kolumn].str.strip() == "").mean()
    print(f"{kolumn:<22} {tomma:.1%} tomma")
```

```python
# Spel som saknar både genrer och beskrivning
tomma_spel = (df_train["genres"].apply(len) == 0) & (df_train["short_description"].str.strip() == "")

print("Andel tomma spel:", f"{tomma_spel.mean():.1%}")
print("Andel hits bland tomma spel:", f"{df_train.loc[tomma_spel, 'hit'].mean():.1%}")
print("Andel hits bland övriga:", f"{df_train.loc[~tomma_spel, 'hit'].mean():.1%}")

# Några exempel
df_train.loc[tomma_spel, ["name", "price", "positive", "hit"]].head(10)
```

**Tomma spel**
- Ungefär 5 % av spelen saknar genrer, kategorier, språk, utvecklare och beskrivning. Det är samma spel i alla kolumner.
- Det är playtester och verktyg (t.ex. "Farming Playtest", "Volcanoids Modding SDK"), inte riktiga spel.
- Ingen av dem är en hit (0 %), jämfört med 26 % av de andra. De ska tas bort när vi rättar datat.
- `tags` är tom för alla spel och kan strykas helt.
- `website` saknas för 56 %, men det är normalt: många spel har ingen egen webbplats.

## Läckage: vilka kolumner får användas?

| Kolumn | Får användas? | Varför |
|---|---|---|
| `positive` | Nej, svaret | Hit räknas från den |
| `negative`, `recommendations`, `reviews`, `user_score`, `score_rank` | Nej, läcker | Recensioner kommer efter släppet |
| `metacritic_score`, `metacritic_url` | Nej, läcker | Betyg kommer efter släppet |
| `estimated_owners`, `peak_ccu` | Nej, läcker | Antal spelare kommer efter släppet |
| alla `playtime`-kolumner | Nej, läcker | Speltid kommer efter släppet |
| `release_date`, `release_year` | Nej | Gamla spel har haft längre tid att få recensioner |
| `appID` | Nej | Bara ett ID-nummer, men lågt ID = gammalt spel, så det smyger in släppåret |
| `tags` | Nej | Tom för alla spel |
| `dlc_count`, `achievements` | Gråzon | Läggs ofta till efter släppet när spelet säljer bra |
| `price`, `packages` | Gråzon | Kan ändras med rea efter släppet |
| `supported_languages`, `full_audio_languages`, `categories` | Gråzon | Kan läggas till efter släppet |
| `genres`, `windows`, `mac`, `linux`, `required_age` | Ja | Bestäms före släppet |
| `short_description`, `detailed_description`, `screenshots`, `header_image` | Ja | Finns på butikssidan innan släppet |
| `developers`, `publishers`, `website`, `support_url`, `support_email`, `notes` | Ja | Finns innan släppet |
| `name`, `movies` | Ja, men hjälper inte | Namnet säger lite, och `movies` är tom |

Gråzon betyder att vi kan använda dem, eftersom Pixel Harbor vet ungefär värdet innan släppet, men värdet i datat kan vara högre än det var vid släppet.

```python
import numpy as np
import matplotlib.pyplot as plt

# Gör om några kolumner till tal (bara för att titta, ändrar inte df_train)
titta = pd.DataFrame({
    "price": df_train["price"],
    "log_price": np.log1p(df_train["price"]),
    "dlc_count": df_train["dlc_count"],
    "log_dlc": np.log1p(df_train["dlc_count"]),
    "achievements": df_train["achievements"],
    "required_age": df_train["required_age"],
    "n_languages": df_train["supported_languages"].apply(len),
    "n_genres": df_train["genres"].apply(len),
    "n_categories": df_train["categories"].apply(len),
    "n_screenshots": df_train["screenshots"].apply(len),
    "desc_length": df_train["detailed_description"].str.len(),
    "windows": df_train["windows"].astype(int),
    "mac": df_train["mac"].astype(int),
    "linux": df_train["linux"].astype(int),
    "hit": df_train["hit"],
})

# Korrelation med hit: nära 0 = inget samband, längre från 0 = starkare
samband = titta.corr()["hit"].drop("hit").sort_values()
samband.plot(kind="barh", figsize=(7, 5), title="Samband med hit")
plt.axvline(0, color="black", linewidth=0.8)
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].hist(df_train["price"], bins=50)
ax[0].set_title("Pris")
ax[1].hist(np.log1p(df_train["price"]), bins=50)
ax[1].set_title("log(1 + pris)")
plt.show()
```

## Vad hänger ihop med hits?

- Alla samband är positiva men ganska svaga (som mest 0,30).
- `n_categories` (antal Steam-funktioner) har starkast samband. Sedan kommer DLC, beskrivningens längd och antal skärmbilder: alltså hur påkostat spelet är.
- **Log-skalan gör stor skillnad.** `dlc_count` har nästan inget samband (0,04), men `log_dlc` har ett av de starkaste (0,29). För priset går det från 0,04 till 0,17. Histogrammet visar varför: utan log ligger nästan alla spel i samma stapel.
- `windows` säger nästan ingenting, eftersom nästan alla spel finns på Windows.
