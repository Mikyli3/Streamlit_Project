```python
import pandas as pd

# Läser in speldatan
df = pd.read_parquet("../data/train-00000-of-00001.parquet")

# Visar antal spel och kolumner
print("Dataset successfully loaded!")
print(f"Number of games: {len(df)}")
print(f"Number of columns: {len(df.columns)}")

# Visar de första fem spelen
df.head()
```

```python
# Visar alla kolumnnamn
print("Dataset columns:")
print(df.columns.tolist())

# Visar vilken typ av data varje kolumn innehåller
print("\nData types:")
print(df.dtypes)

# Kollar om värden saknas
print("\nMissing values:")
print(df.isnull().sum().sort_values(ascending=False))

# Visar information om datan
print("\nDataset information:")
df.info()
```

```python
# Kollar om samma spel-ID finns flera gånger
print("Duplicate appIDs:")
print(df["appID"].duplicated().sum())

# Visar saknade värden
print("\nMissing values:")
missing = df.isnull().sum().sort_values(ascending=False)
display(missing[missing > 0])

# Visar information om datan
print("\nDataset information:")
df.info()

# Visar enkel statistik
print("\nNumerical statistics:")
display(df.describe())
```

```python
# Visar kolumner och datatyper
print("ALL DATASET COLUMNS\n")

for column in df.columns:
    print(f"{column:<35} {df[column].dtype}")

# Visar exempel från viktiga kolumner
important_columns = [
    "name",
    "release_date",
    "estimated_owners",
    "peak_ccu",
    "price",
    "metacritic_score",
    "user_score",
    "positive",
    "negative",
    "recommendations",
    "average_playtime_forever",
    "genres",
    "categories"
]

print("\nEXAMPLE DATA\n")
display(df[important_columns].head(10))
```

```python
# Väljer kolumner att undersöka
columns_to_check = [
    "name",
    "release_date",
    "estimated_owners",
    "positive",
    "negative",
    "recommendations",
    "genres"
]

# Visar 15 slumpmässiga spel
print("SAMPLE OF 15 GAMES")
display(
    df[columns_to_check].sample(15, random_state=42)
)

# Visar exempel på släppdatum
print("\nRELEASE DATE EXAMPLES")
print(df["release_date"].head(20).to_list())

# Visar uppskattat antal ägare
print("\nESTIMATED OWNERS EXAMPLES")
print(df["estimated_owners"].value_counts().head(15))

# Visar statistik om recensioner
print("\nREVIEW STATISTICS")
display(df[["positive", "negative"]].describe())
```
